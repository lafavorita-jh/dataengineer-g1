```python
from datetime import datetime
import time

total_intentos = 3
n_intento = 0
errores = []

while n_intento < total_intentos:
    try:
        path = "../data/notas.txt"
        with open(path, "r") as f:
            contenido = f.read()
            print(contenido)
            break
        n_intento += 1
    except FileNotFoundError as e:
        errores.append(f"{datetime.now()} | Error: {e}")
        with open("../data/errores.txt", "w") as f:
           f.write("\n".join(errores))

        n_intento += 1
        time.sleep(10)
print(errores)
```
