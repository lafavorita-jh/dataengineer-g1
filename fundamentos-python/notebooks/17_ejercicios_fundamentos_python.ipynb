### Pregunta 1.2: Ordenar palabras por longitud

Dada una lista de palabras, ordénalas desde la **más corta** hasta la **más larga**.
Luego muestra también el orden inverso (de la más larga a la más corta).

```python
palabras = ['casa', 'elefante', 'sol', 'computadora', 'mesa', 'programacion']
```

```python
# Escribe tu solución aquí
# Sol Corta
palabras.sort() 
palabras.sort(key=len)
print(f"Palabras ordenadas por longitud de más corta a mas larga: \n {palabras}")
palabras.sort(key=len,reverse=True)
print(f"Palabras ordenadas por longitud inversamente: \n {palabras}")

#Sol larga
diccionario_palabras = {} # Crear diccionario
for p in palabras: # Recorrer la lista con un for y asignar palabra: longitud
    diccionario_palabras[p] = len(p)
ordenado=[]
while len(diccionario_palabras)>0:
    palabra_menor = min(diccionario_palabras, key=diccionario_palabras.get)
    ordenado.append(palabra_menor)
    del diccionario_palabras[palabra_menor]
# Resultado
print("_____________")
print(f'Las palabras ordenadas acendentemente son: \n {ordenado}')

print(f'Las palabras ordenadas decendente son:\n {ordenado[::-1]}')
```

### Pregunta 3.1: Consolidación de inventarios

Combina tres inventarios provenientes de distintas sucursales. Cuando un producto aparezca en varias
sucursales, **suma sus cantidades** y calcula también **en cuántas sucursales** está disponible.

**Requisitos:**
- Crea un diccionario final donde cada producto tenga las claves `cantidad_total` y `sucursales`.
- No modifiques los diccionarios originales.
- Ordena el resultado alfabéticamente por producto.

```python
inventario_norte = {"laptop": 4, "mouse": 10, "monitor": 3}
inventario_sur = {"laptop": 2, "teclado": 8, "monitor": 5}
inventario_centro = {"mouse": 6, "teclado": 4, "webcam": 7}
```

```python
# Escribe tu solución aquí
inventario_total = {}
articulos=set([
    *inventario_norte.keys(),
    *inventario_sur.keys(),
    *inventario_centro.keys()
])
print(articulos)
inventarios= [inventario_norte, inventario_sur, inventario_centro]
for articulo in articulos:
    inventario_total[articulo]={
        'cantidad_total': sum([inventario.get(articulo, 0) for inventario in inventarios]),
        'sucursales': sum(1 for inventario in inventarios if articulo in inventario),
    }
print(inventario_total)
for articulo in inventario_total:
    print(f"Articulo: {articulo}, Cantidad total: {inventario_total[articulo]['cantidad_total']}, Sucursales: {inventario_total[articulo]['sucursales']}")
```

### Pregunta 3.3: Agrupación jerárquica de empleados

Agrupa a los empleados primero por **departamento** y luego por **nivel** de experiencia.

**Requisitos:**
- En cada grupo almacena `nombres`, `Cantidad` y `salario_promedio`.
- La estructura debe ser un diccionario anidado: `{departamento: {nivel: {...}}}`.

```python
empleados = [
    {"nombre": "Ana", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5200},
    {"nombre": "Luis", "departamento": "Diseño", "nivel": "Junior", "salario": 2800},
    {"nombre": "Carla", "departamento": "Desarrollo", "nivel": "Junior", "salario": 3400},
    {"nombre": "Pedro", "departamento": "Desarrollo", "nivel": "Senior", "salario": 5600},
    {"nombre": "Lucía", "departamento": "Diseño", "nivel": "Senior", "salario": 4700},
]
```

```python
# Escribe tu solución aquí
departamentos = list(set([
    empleado["departamento"] for empleado in empleados
]))
experiencia = list(set([
    empleado["nivel"] for empleado in empleados
]))
print(f'Departamentos: {departamentos}')
print(f'Experiencia: {experiencia}')
#Agrupameinto pr departamento
resultado = {}

for dep in departamentos:
    resultado[dep] = {}
    for exp in experiencia:
        # Filtrar empleados que pertenecen al mismo departamento y nivel
        grupo = [e for e in empleados if e["departamento"] == dep and e["nivel"] == exp]
        print(f'Grupo: {grupo}')
        if grupo: #Si existen valores en el grupo, entonces calcular nombres, cantidad y salario promedio
            nombres = [e["nombre"] for e in grupo]
            cantidad = len(grupo)
            salario_promedio = sum(e["salario"] for e in grupo) / cantidad

            # Almacenar nombres, Cantidad y salario_promedio
            resultado[dep][exp] = {
                "nombres": nombres,
                "Cantidad": cantidad,
                "salario_promedio": salario_promedio
            }
for dep, niveles in resultado.items():
    print(f'Departamento: {dep}')
    for exp, datos in niveles.items():
        print(f'  Nivel: {exp}')
        print(f'    Nombres: {datos["nombres"]}')
        print(f'    Cantidad: {datos["Cantidad"]}')
        print(f'    Salario Promedio: {datos["salario_promedio"]:.2f}')

```

### Pregunta 5.1: Analizar las ventas de una tienda

Crea una función `analizar_ventas(ventas)` que reciba una lista de montos de venta y construya un resumen estadístico.

**Requisitos:**
- Calcular el total vendido, el promedio, la venta más alta y la venta más baja.
- Contar cuántas ventas fueron superiores al promedio.
- Indicar cuántas ventas se encuentran por debajo de 100.
- Retornar todos los resultados dentro de un diccionario.
- Si la lista está vacía o contiene valores negativos, retornar un mensaje de error.

**Reto adicional:** agrega un parámetro opcional `meta`. La función debe indicar si el total vendido alcanzó dicha meta.

```python
ventas = [150.50, 200, 80.90, 300, 175.50, 90]
```

```python
# Escribe tu solución aquí
def analizar_ventas(ventas: list[float]) -> dict[str, float] | str:
    if not ventas:
        return 'ERROR: No existen ventas para analizar'
    for venta in ventas:
        if venta < 0:
            return f'ERROR: No se permiten ventas negativas: {venta}'
    total_vendido=sum(ventas)
    promedio_ventas=total_vendido/len(ventas)
    venta_maxima=max(ventas)
    venta_minima=min(ventas)
    ventas_superiores_promedio=sum([1 for venta in ventas if (venta>promedio_ventas)])
    ventas_inferiores_100=sum([1 for i in ventas if (i<100)])
    estadisticas = {
        "total_vendido": total_vendido,
        "promedio_ventas": promedio_ventas,
        "venta_maxima": venta_maxima,
        "venta_minima": venta_minima,
        "ventas_superiores_promedio": ventas_superiores_promedio,
        "ventas_inferiores_100": ventas_inferiores_100,
    }
    return estadisticas

stats =analizar_ventas(ventas)
print(f'{key}: {value}' for key, value in stats.items())
```

- Calcular el promedio de cada empleado.
- Clasificar el resultado como **Excelente** (≥ 90), **Bueno** (≥ 80), **Regular** (≥ 70) o **Deficiente** (< 70).
- Incluir únicamente a quienes tengan un promedio igual o superior a `minimo`.
- Retornar una lista de tuplas con `(nombre, promedio, clasificación)`.
- Ordenar de mayor a menor promedio y, en caso de empate, alfabéticamente.

```python
empleados = {
    'Ana': [80, 90, 75],
    'Carlos': [60, 70, 65],
    'Lucía': [95, 85, 90],
    'Pedro': [70, 70, 70],
}
```

```python
# Escribe tu solución aquí

def evaluar_empleados(empleados, minimo=70)-> list[tuple[str, float, str]]:
    resultado = []
    for nombre, puntajes in empleados.items():
        promedio = sum(puntajes) / len(puntajes)
        if promedio >= minimo:
            if promedio >= 90:
                clasificacion = "Excelente"
            elif promedio >= 80:
                clasificacion = "Bueno"
            elif promedio >= 70:
                clasificacion = "Regular"
            else:
                clasificacion = "Deficiente"
            resultado.append((nombre, round(promedio, 2), clasificacion))

    resultado.sort(key=lambda x: (-x[1], x[0]))

    return resultado



# Ejecución
evaluados = evaluar_empleados(empleados, minimo=70)
print(evaluados)
```

  Por ejemplo, `aaabbccccdaa` debe convertirse en `a3b2c4d1a2`.
- `descomprimir_texto` debe reconstruir el texto original.
- La descompresión debe admitir cantidades de más de un dígito, como `a12`.
- Si el formato comprimido es inválido, retornar un mensaje de error.

```python
texto = 'aaaaaaaaaaaaaaaaabbccccdaa'
```

```python
# Escribe tu solución aquí

def comprimir_texto(texto: str) -> str:
    resultado = []
    char_actual = texto[0]
    conteo = 1
    for c in texto[1:]:
        if c == char_actual:
            conteo += 1
        else:
            resultado.append(f"{char_actual}{conteo}")
            char_actual = c
            conteo = 1   
    resultado.append(f"{char_actual}{conteo}")
    
    return "".join(resultado)


prueba=comprimir_texto(texto)
print(f'Texto comprimido: {prueba}')
def descomprimir_texto(texto: str) -> str:
    if not texto:
        return ""
    
    resultado = []
    char = ""
    cant = ""
    
    for c in texto:
        if not c.isdigit():
            if char:
                resultado.append(char * int(cant))
                cant = ""
            char = c
        else:
            cant += c
            
    # Agrega el último grupo guardado
    resultado.append(char * int(cant))
    
    return "".join(resultado)
a=descomprimir_texto(prueba)
print(f'Texto descomprimido: {a}')
```

```python
sala = [['L', 'L', 'O', 'L'],
        ['O', 'L', 'L', 'L'],
        ['L', 'O', 'L', 'L']]
asientos = [(1, 2), (2, 3), (4, 1)]
```

```python
# Escribe tu solución aquí
def reservar_asientos(sala, asientos_solicitados):
    try:
        for fila, col in asientos_solicitados:
            if fila < 1 or col < 1:
                return False, f"Error: Las posiciones deben ser mayores a 0. Posición inválida: ({fila}, {col})", sala
            r, c = fila - 1, col - 1
            asiento = sala[r][c]
            if asiento != 'L':
                return False, f"Error: El asiento ({fila}, {col}) ya está ocupado.", sala
        for fila, col in asientos_solicitados:
            r, c = fila - 1, col - 1
            sala[r][c] = 'O'
        return True, asientos_solicitados
    except IndexError:

        return False, "Error: Una o más posiciones solicitadas están fuera del rango de la sala.", sala
    except (TypeError, ValueError):
       
        return False, "Error: El formato de los datos ingresados es inválido.", sala

    except Exception as e:
          return False, f"Error inesperado: {str(e)}", sala


def mostrar_sala(sala):
    if not sala:
        print("La sala está vacía.")
        return
    num_cols = len(sala[0])
    encabezado = "     " + " ".join(f"C{c + 1}" for c in range(num_cols))
    linea_separadora = "   " + "-" * (len(encabezado) - 3)
    print(encabezado)
    print(linea_separadora)

    for idx, fila in enumerate(sala):
        asientos_str = "  ".join(fila)
        print(f"F{idx + 1} |  {asientos_str}")
a=mostrar_sala(sala)
print(a)
b=reservar_asientos(sala, asientos)
print(b)
```

- Método `consultar_saldo()`: devuelve el saldo actual.

Prueba la clase con una cuenta: deposita, retira un monto válido y luego intenta retirar más de lo disponible.

*Objetivo: modificar el estado interno del objeto con validaciones.*

```python
# Escribe tu solución aquí
class CuentaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = saldo_inicial

    def depositar(self, cantidad)-> str:
        if cantidad <= 0:
            return "Error: La cantidad a depositar debe ser positiva."
        self.saldo += cantidad
        return f"Depósito exitoso. Nuevo saldo: {self.saldo}"

    def retirar(self, cantidad) -> str:
        if cantidad <= 0:
            return "Error: La cantidad a retirar debe ser positiva."
        if cantidad > self.saldo:
            return "Error: Fondos insuficientes."
        self.saldo -= cantidad
        return f"Retiro exitoso. Nuevo saldo: {self.saldo}"

    def consultar_saldo(self) -> str:
        return f"Saldo actual: {self.saldo}"

cuenta_01=CuentaBancaria("Ana", 1000)
print(cuenta_01.titular)
print(cuenta_01.consultar_saldo())
print(cuenta_01.depositar(500))
print(cuenta_01.retirar(2000))
print(cuenta_01.consultar_saldo())
```

### Pregunta 6.5: Sistema de biblioteca (ejercicio integrador)

Crea tres clases que trabajen juntas: `Libro`, `Usuario` y `Biblioteca`.

- `Libro`: `titulo`, `autor` y `disponible` (`True` por defecto). *(Es una clase distinta a la del ejercicio 6.2: puedes redefinirla.)*
- `Usuario`: `nombre` y una lista de `libros_prestados`.
- `Biblioteca`: guarda una lista de libros y ofrece los métodos `agregar_libro()`, `buscar(titulo)`,
  `prestar(titulo, usuario)`, `devolver(titulo, usuario)` y `catalogo()`.
- Al prestar un libro debe marcarse como no disponible y añadirse a la lista del usuario; al devolverlo, lo contrario.
- Contempla los casos: el libro no existe, el libro ya está prestado, el usuario no tiene ese libro.

*Objetivo: integrar varias clases que se relacionan entre sí, tal como ocurre en un programa real.*

```python
# No hay datos de entrada: crea tus propios objetos para probar las clases.
```

```python
class Libro:
    def __init__(self, titulo: str, autor: str, disponible: bool = True):
        self.titulo = titulo
        self.autor = autor
        self.disponible = disponible

    def __str__(self) -> str:
        estado = "Disponible" if self.disponible else "Prestado"
        return f"'{self.titulo}' de {self.autor} [{estado}]"


class Usuario:
    def __init__(self, nombre: str, libros_prestados: list = None):
        self.nombre = nombre
        self.libros_prestados = libros_prestados if libros_prestados is not None else []


class Biblioteca:
    def __init__(self):
        self.libros = []

    def agregar_libro(self, libro: Libro):
        self.libros.append(libro)
        print(f"Libro '{libro.titulo}' agregado a la biblioteca.")

    def buscar(self, titulo: str):
        for libro in self.libros:
            if libro.titulo.lower() == titulo.lower():
                return libro
        return None

    def prestar(self, titulo: str, usuario: Usuario):
        libro = self.buscar(titulo)

        if libro is None:
            print(f"Error: El libro '{titulo}' no existe en la biblioteca.")
            return

        if not libro.disponible:
            print(f"Error: El libro '{titulo}' ya se encuentra prestado.")
            return

        libro.disponible = False
        usuario.libros_prestados.append(libro)
        print(f"Éxito: Se prestó '{libro.titulo}' a {usuario.nombre}.")

    def devolver(self, titulo: str, usuario: Usuario):
        libro = self.buscar(titulo)

        if libro is None:
            print(f"Error: El libro '{titulo}' no pertenece a esta biblioteca.")
            return

        if libro not in usuario.libros_prestados:
            print(f"Error: El usuario {usuario.nombre} no tiene prestado el libro '{titulo}'.")
            return

        libro.disponible = True
        usuario.libros_prestados.remove(libro)
        print(f"Éxito: {usuario.nombre} ha devuelto el libro '{libro.titulo}'.")

    def catalogo(self):
        print("\n--- CATÁLOGO DE LA BIBLIOTECA ---")
        if not self.libros:
            print("La biblioteca está vacía.")
            return
        for libro in self.libros:
            print(f"- {libro}")
        print("---------------------------------\n")



biblio = Biblioteca()

l1 = Libro("Cien años de soledad", "Gabriel García Márquez")
l2 = Libro("Don Quijote", "Miguel de Cervantes")
biblio.agregar_libro(l1)
biblio.agregar_libro(l2)

u1 = Usuario("Carlos")
u2 = Usuario("Ana")

biblio.catalogo()

biblio.prestar("Cien años de soledad", u1)  # Préstamo exitoso
biblio.prestar("Cien años de soledad", u2)  #  Ya está prestado
biblio.prestar("Rayuela", u1)              # El libro no existe

biblio.catalogo()

biblio.devolver("Cien años de soledad", u2)  #  Ana no tiene el libro
biblio.devolver("Cien años de soledad", u1)  # Devolución exitosa

biblio.catalogo()
```
