```python
codigo='V-2024-1001'
print(f"El codigo es: {codigo}")

print(f"las posiciones 1 a 4 del codigo son: {codigo[1:4]}")
print(f"El codigo invertido es: {codigo[::-1]}")
```

```python
#Metodos de limpieza
nombre= '  Jhunior         avila'
apellido_materno= 'Quintero'
print(f"El nombre es: {nombre.strip()}")#Quita espacios al inicio y final
print(f"El codigo en mayusculas es: {nombre.upper()}")
print(nombre.title())#Convierte la primera letra de cada palabra en mayuscula
print(nombre.replace(" ","").replace("RA","r A").title())
```

```python
import re
nombre="  Ana   TORRES        "
nombre_limpio=re.sub(r"\s+"," ",nombre).strip().title()
print(f"El nombre limpio es: {nombre_limpio}")
```
