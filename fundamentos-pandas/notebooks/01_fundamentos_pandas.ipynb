##  Data frames
### Crear Dataframes desde un diccionario

```python
df=pd.DataFrame({
    'producto': ['Laptop','Monitor','Teclado','Mouse'],
    'unidades': [10,20,30,40],
    'precio': [100,200,300,400],
})
```

```python
print(df)
```

### dESDE UNA LISTA DE DICCIONARIOS

```python
### dESDE UNA LISTA DE DICCIONARIOS
df2=pd.DataFrame([
    {'producto': 'Laptop', 'unidades': 10, 'precio': 100},
    {'producto': 'Monitor', 'unidades': 20, 'precio': 200},
    {'producto': 'Teclado', 'unidades': 30, 'precio': 300},
    {'producto': 'Mouse', 'unidades': 40, 'precio': 400}
])
```

```python
print(df2)
```

### Desde una lista de listas

```python
df3=pd.DataFrame(
    [
    ['Laptop', 10, 100],
    ['Monitor', 20, 200],
    ['Teclado', 30, 300],
    ['Mouse', 40, 400]
], 
columns=['producto', 'unidades', 'precio']
)
df3
```

### Operaciones


```python
# Columnas calculadas
df3['Inv_Valorizado'] = df3['unidades'] * df3['precio']
print(df3)
df3.shape
print(df3.columns.to_list())
print(df3.describe())
```
