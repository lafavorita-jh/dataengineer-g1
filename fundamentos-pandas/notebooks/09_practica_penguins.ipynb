### 21. Crear una nueva columna

Crea una columna llamada `body_mass_kg` con la masa corporal expresada en kilogramos. Recuerda: 1000 g = 1 kg. No modifiques `body_mass_g`.

````python
# Escribe tu solución aquí
df['body_mass_kg']=df['body_mass_g']*1000
print(df.columns.to_list())
````

### 22. Clasificación mediante pd.cut()

Crea una nueva columna `mass_category` clasificando los pingüinos como **Light** si pesan menos de 3500 g, **Medium** si pesan entre 3500 y 4500 g (inclusive) y **Heavy** si pesan más de 4500 g. Utiliza `pd.cut()`.

````python
# Escribe tu solución aquí
df['mass_category']=pd.cut(
    df['body_mass_g'],
    bins=[-float('inf'),3500,4500,+float('inf')],
    labels=['Light','Medium','Heavy'],
)
````

````python
df[df['body_mass_g']<=3500]
````

### 23. Masa promedio por especie

Calcula la masa corporal promedio (`body_mass_g`) de cada especie. El resultado debe tener una fila por especie.

````python
# Escribe tu solución aquí
masa_promedio=df.groupby('species')['body_mass_g'].agg(masa_primedio_g='mean',desviacion_g='std')
masa_promedio
````

### 24. Múltiples estadísticas

Para cada especie calcula sobre `body_mass_g`: cantidad, promedio, mínimo, máximo y desviación estándar. Intenta resolverlo utilizando `groupby()` y `agg()`.

````python
# Escribe tu solución aquí
agrupado=df.groupby('species')['body_mass_g'].agg(
    cantidad='count',
    promedio='mean',
    minimo='min',
    maximo='max',
    desviacio='std',
)
agrupado
````

### 25. Agrupación utilizando dos columnas

Calcula la masa corporal promedio agrupando por `species` y `sex`. El resultado debe permitir comparar machos y hembras dentro de cada especie.

````python
# Escribe tu solución aquí
agrupado=df.groupby(['species','sex']).agg({
    'body_mass_g':['mean']
})
agrupado
````

### 26. Tabla cruzada

Construye una tabla que muestre cuántos pingüinos existen para cada combinación `species` × `island`. Puedes utilizar `pd.crosstab()` o `pivot_table()`.

### 27. Pingüino más pesado de cada especie

Encuentra el registro correspondiente al pingüino con mayor `body_mass_g` dentro de cada especie. El resultado final debe mostrar `species`, `sex`, `island` y `body_mass_g`. No basta con mostrar únicamente la masa máxima: debes recuperar el registro completo correspondiente al máximo.

````python
# Escribe tu solución aquí
mas_pesado=df.groupby(['species','sex','island']).agg({
    'body_mass_g':['max']
})
mas_pesado
````

### 28. Diferencia respecto al promedio

Crea una columna `mass_difference` que indique cuánto pesa cada pingüino por encima o por debajo del promedio de su propia especie. Conceptualmente: *masa del pingüino − masa promedio de su especie*. Intenta resolverlo usando `groupby()` y `transform()`.

````python
# Escribe tu solución aquí


# 1. Calcular el promedio de masa corporal por especie y propagarlo a cada fila
promedio_por_especie = df.groupby("species")["body_mass_g"].transform("mean")

# 2. Restar la masa individual del promedio de su especie
df["mass_difference"] = df["body_mass_g"] - promedio_por_especie

# Mostrar las columnas relevantes para verificar el resultado
print(df[["species", "body_mass_g", "mass_difference"]].head())
````

### 29. Ranking dentro de cada especie

Crea una columna `mass_rank` que indique el ranking de masa corporal de cada pingüino dentro de su propia especie. El pingüino más pesado debe tener `mass_rank = 1`. Luego muestra los 3 pingüinos más pesados de cada especie.

````python
df.columns.to_list()
````

````python
# Escribe tu solución aquí
#Con group by y rank
df['mass_rank']=df.groupby('species')['body_mass_g'].rank(method='first',ascending=True)
````

````python
df[df['mass_rank']<=3].sort_values(['species','mass_rank'])
````
