### LOC

```python
### LOC: Localiza las filas por indice, dado que las columnas no tienen indice se debe usar su etiqueta o nombre
```

```python
ventas_df.loc[0: 4, ["vendedor", "producto"]]
```

```python
#Muestra las primeras 1000 filas de las columnas vendedor y producto
ventas_df.loc[:1000, ["vendedor", "producto"]]
```

Iloc

```python
ventas_df.iloc[0: 4, [0, 1]]
```

```python
print(ventas_df.dtypes)
# 1. Obtener los valores únicos de cada columna
vendedores = ventas_df["vendedor"].unique()
productos = ventas_df['producto'].unique()
categorias = ventas_df['categoria'].unique()
regiones = ventas_df['region'].unique()

# 2. Imprimir los resultados
print("Vendedores:", list(vendedores))
print("Productos:", list(productos))
print("Categorías:", list(categorias))
print("Regiones:", list(regiones))
```
